# BrainrotVision EDA Overview

This notebook reads the generated metadata and EDA outputs produced by the shared Python pipeline.

Run `make metadata`, `make eda`, and `make artifacts` first so the plots and tables exist on disk.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

data_dir = Path('../data/processed')
stats = json.loads((data_dir / 'dataset_stats.json').read_text())
metadata = pd.read_parquet(data_dir / 'metadata.parquet')
stats, metadata.head()
```

```python
plots_dir = data_dir / 'plots'
for plot_name in [
    'format_distribution.png',
    'width_distribution.png',
    'height_distribution.png',
    'aspect_ratio_distribution.png',
    'brightness_distribution.png',
    'contrast_distribution.png',
    'sample_gallery.png',
    'embedding_projection.png',
]:
    plot_path = plots_dir / plot_name
    if plot_path.exists():
        display(Image(filename=str(plot_path)))
```
